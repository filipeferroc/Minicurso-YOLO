```python
import os

from ultralytics import YOLO
```

* **Definir o diretório**

    Importante para gravar e visualizar a predição


```python
# CAMINHO DO DIRETÓRIO ESCOLHIDO

project = '/content/drive/MyDrive/Faculdade/Minicurso_YOLO/Images_teste'

path = "/content/drive/MyDrive/Faculdade/Minicurso_YOLO"
```

```python
# Imagem

name_img = 'violao.jpg'
```

```python
# Diretório para guardar a predição

subpath_predict = 'predicoes-3'
```

* **Carregar o modelo 11 do YOLO**

```python
# Load a model
model = YOLO("yolo11n.pt")
```

```python
# CAMINHO DA IMAGEM PARA REALIZAR A PREDIÇÃO

image_path = '/content/drive/MyDrive/Faculdade/Minicurso_YOLO/Images_teste/' + name_img


# FAZER A PREDIÇÃO

results = model.predict(
    source=image_path,
    save=True,
    project=project,          # Define a pasta base no Drive
    name=subpath_predict      # Cria a subpasta predicoes
)
```

```python
for result in results:
  boxes = result.boxes
  result.save(filename='result.jpg')
```

```python
import os

# O YOLO salvou dentro de 'project' usando o prefixo do 'subpath_predict'
# Como você rodou mais de uma vez, ele gerou sufixos como 'predicoes-3'.
# Vamos listar a pasta do projeto para ver as pastas reais de predição:
print("Conteúdo da pasta do projeto:")
print(os.listdir(project))
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

path_predict = project + '/' + subpath_predict + '/' + name_img

img = Image.open(os.path.join(path, path_predict))
image = mpimg.imread(path_predict)

plt.figure(figsize=(10, 10))
plt.imshow(img)
plt.axis('off')
plt.show()
```
